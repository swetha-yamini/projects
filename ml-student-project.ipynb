# *Data Preprocessing*
### ***Encode categorical variables (Label Encoding).***

```python
categorical_cols = df.select_dtypes(include='object').columns.tolist()
categorical_cols.remove('Target') 
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoders = {}
# Encode and store each encoder
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])      # convert strings to numbers
    label_encoders[col] = le                 # store encoder
```

```python
target_encoder = LabelEncoder()
df['Target'] = target_encoder.fit_transform(df['Target'])
```

### ***Predict Function(Label-Encoded Features)***

```python
def predict_student_label_encoded(student_input: dict, model, label_encoders, target_encoder):
    input_df = pd.DataFrame([student_input])
    # Encode categorical features
    for col in input_df.columns:
        if col in label_encoders:
            le = label_encoders[col]
            try:
                input_df[col] = le.transform(input_df[col])
            except ValueError:
                print(f"Value '{input_df[col].values[0]}' not seen in training for column '{col}'")
                return None
    # Ensure column order matches training
    input_df = input_df[model.feature_names_in_]
    # Predict
    prediction = model.predict(input_df)
    return target_encoder.inverse_transform(prediction)[0]
```

```python
student_input ={'Marital status': 'Single',
    'Application mode': 'Online - Website',
    'Application order': '1st Preference',
    'Course': 'B.Sc. Biology',
    'Daytime/evening attendance': 'Daytime',
    'Previous qualification': 'No Formal Education',
    'Debtor': ' Not a Debtor',
    'Tuition fees up to date': 'Up to Date',
    'Gender': 'Female',
    'Scholarship holder': 'No Scholarship',
    'Age at enrollment': 18,
    'Curricular units 1st sem (enrolled)': 6,
    'Curricular units 1st sem (evaluations)': 6,
    'Curricular units 1st sem (approved)': 6,
    'Curricular units 1st sem (grade)': 14.0,
    'Curricular units 2nd sem (enrolled)': 6,
    'Curricular units 2nd sem (evaluations)': 6,
    'Curricular units 2nd sem (approved)': 6,
    'Curricular units 2nd sem (grade)': 13.5,
    'Unemployment rate': 8.0,
    'GDP': 1.5
}
result = predict_student_label_encoded(student_input, xg, label_encoders, target_encoder)
print("Predicted Status:", result)
```
