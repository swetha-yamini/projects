# <p style="font-family:newtimeroman;font-size:150%;text-align:center;color:#06445e;">Correlation Analysis:</p>

```python
correlation_matrix = df.corr()
print(correlation_matrix)
# Create a heatmap to visualize the correlations
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix (Pearson)")
plt.show()
# Calculate Spearman correlation coefficients (for non-linear relationships)
spearman_correlation_matrix = df.corr(method="spearman")
print(spearman_correlation_matrix)
# Create a heatmap to visualize the Spearman correlations
sns.heatmap(spearman_correlation_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix (Spearman)")
plt.show()
```

```python
# Calculate correlation coefficient between confirmed and Recovered
correlation_coefficient = np.corrcoef(df['Confirmed'], df['Recovered'])[0, 1]
print(correlation_coefficient)
```

#### **<h1 align="center"><span style="color:#06445e;">Calculate summary statistics for groups</span>**

```python
# Mean rating by category
mean_sales_by_category = df.groupby('Country/Region')['Active'].mean()
print(mean_sales_by_category)

# Median rating by review_content
median_sales_by_age = df.groupby('New cases')['New deaths'].median()
print(median_sales_by_age)

# Standard deviation of actual_price by product_name
std_price_by_brand = df.groupby('WHO Region')['Confirmed'].std()
print(std_price_by_brand)
```

# <p style="font-family:newtimeroman;font-size:150%;text-align:center;color:#06445e;">Statistical Tests:</p>

```python
import scipy.stats as stats
# Conduct t-test to compare rating between two categories
t_statistic, p_value = stats.ttest_ind(df[df['Country/Region'] == 'Recovered']['Confirmed'], df[df['Country/Region'] == 'Deaths']['Confirmed'])
print(t_statistic, p_value)
```

```python
# Chi-square test
contigency_table = pd.crosstab(df['WHO Region'], df['Deaths'])
contigency_table
```

```python
# perform chi-square test
chi2, p, dof, expected = stats.chi2_contingency(contigency_table)
print('Chi-square statistic:', chi2)
print('p-value:', p)
print('Degrees of freedom:', dof)
print(f"Expected:\n {expected}")
```

# <p style="font-family:newtimeroman;font-size:150%;text-align:center;color:#06445e;">Questions and Answers</p>
### ***These are some questions are follows:***

### ***Q1: Which country has the highest number of confirmed cases?***

```python
country_max_confirmed = df.loc[df['Confirmed'].idxmax(), ['Country/Region', 'Confirmed']]
print(country_max_confirmed)
```

### ***Answer 1:*** 
- ***The highest number of confirmed cases is United States (US) with 4,290,259 cases***
- The US has the highest absolute number of COVID-19 cases, likely due to:
- High testing rates
- Population size
- Early widespread community transmission
- The large case volume places major strain on healthcare resources.

### ***Q2: What is the average death rate (Deaths / 100 Cases) across all countries?***

```python
avg_death_rate = df['Deaths / 100 Cases'].mean()
print(f"{avg_death_rate:.2f}%")
```

### ***Answer 2:*** 
- ***The average death rate across all countries is 3.02%***
- This is a global average, and real fatality rates vary significantly:
- Higher in countries with older populations (e.g., Italy)
- Lower in those with strong public health measures and early interventions (e.g., South Korea)

### ***Q3: Which WHO Region has the highest total confirmed cases?***

```python
region_confirmed = df.groupby('WHO Region')['Confirmed'].sum().sort_values(ascending=False)
top_region = region_confirmed.head(1)
print(top_region)
```

### ***Answer 3:*** 
- ***The highest total confirmed cases is Americas with 8,839,286 cases***
- The Americas (North, Central, and South) had the largest regional outbreak, with Brazil and the US leading. Possible reasons:
- Delayed lockdowns
- Uneven healthcare access
- High population mobility

### ***Q4: Which country had the highest percentage increase in confirmed cases over the last week?***

```python
country_max_percent_increase = df.loc[df['1 week % increase'].idxmax(), ['Country/Region', '1 week % increase']]
print(country_max_percent_increase)
```

### ***Answer 4:*** 
- ***The highest percentage increase in confirmed cases over the last week is Papua New Guinea with a 226.32% increase***
- A sharp weekly increase often indicates:
- Emerging outbreaks
- New community transmission
- Increased testing after long underreporting

### ***Q5: Which country has the highest death rate per 100 recovered patients?***

```python
country_max_death_per_recovered = df.loc[df['Deaths / 100 Recovered'].idxmax(),['Country/Region', 'Deaths / 100 Recovered']]
print(country_max_death_per_recovered)
```

Canada, with an infinite value (likely due to 0 recovered cases or data issues causing a division by zero).***
